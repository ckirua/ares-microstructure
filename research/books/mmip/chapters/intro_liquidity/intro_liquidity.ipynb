# Intro — liquidity features (desk scan)

Loads `research/books/mmip/out/intro_liquidity/exp_intro_eth_summary.json` and plots role-blur / spreads.
Helpers: `research.lib`.

```python
from pathlib import Path
import json, sys
import matplotlib.pyplot as plt

BOOK_ROOT = Path('../..').resolve()
ROOT = BOOK_ROOT
sys.path.insert(0, str(BOOK_ROOT.parents[2]))
from research.lib import fei  # noqa: F401 — smoke import

p = BOOK_ROOT / 'out' / 'intro_liquidity' / 'exp_intro_eth_summary.json'
d = json.loads(p.read_text())
panel = d['features']['panel']
per = d['features']['per_venue']
venues = list(per)
spreads = [per[v]['mean_spread_bps'] for v in venues]
blur = [panel['role_blur_l1'][v] for v in venues]
fig, ax = plt.subplots(1, 2, figsize=(9, 3.5))
ax[0].bar(venues, spreads)
ax[0].set_title('Mean quoted spread (bps)')
ax[1].bar(venues, blur)
ax[1].set_title('Role-blur L1 |upd − size|')
plt.tight_layout()
print('FEI size/updates', panel['fei_size'], panel['fei_updates'])
print('spread stability', panel['spread_stability'])
```
